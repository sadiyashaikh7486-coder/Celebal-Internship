```python
silver_table = spark.table("silver_transactions")

total_records = silver_table.count()

null_txn_ids = (
    silver_table
    .filter("txn_id IS NULL")
    .count()
)

duplicate_txn_ids = (
    total_records -
    silver_table.dropDuplicates(["txn_id"]).count()
)

invalid_amount_records = (
    silver_table
    .filter("amount <= 0")
    .count()
)

print("========== DATA QUALITY REPORT ==========")
print(f"Total Records           : {total_records}")
print(f"Null Transaction IDs    : {null_txn_ids}")
print(f"Duplicate Transaction IDs : {duplicate_txn_ids}")
print(f"Invalid Amount Records  : {invalid_amount_records}")
print("=========================================")
```
